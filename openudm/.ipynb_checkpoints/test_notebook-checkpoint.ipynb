# 1. Multi Criteria Evaluation
* Read from Attractor Table the attractor layer file name and reverse polarity flag.
* For each attractor layer, load file and read raster, standardise acccording to reverse polarity flag. 
    
    * Store the standardised attractor layers to the output path.
* Use **RasterToolkit** module's **RasteriseAreaThresholds** function,

    * Load file and read from the constaint table constrain layer name, dev_flag, layer_threshold.
    * Load constraint rasters' data as a list - inputCoverage - of 2d numpy arrays
    * Calculate summedThresholdArea (m^2) = coverage_threshold (from parameter table, a percentage)/cellsize^2 
    * Calculate layerThresholdArea: for each constraint layer, perform the above equation with its layerThreshold parameter
    * Calculate summedLayerArea by adding element-wise constraint layer arrays, i.e. adding the area of different kinds of constraints.
    * Initiate the final constraint layer with an array of 1. Do element-wise check:

        * If any constraint layer's element > the layer's threshold, meaning the cell's suitability is 0, assign the corresponding element in final constraint array to 0.
        * If the sum of all constraint layers' value at the cell > summedThresholdArea, also assign 0.
    * Write the output, which is the constraint/suitability layer.

    * **Current development layer**: if a layer's development flag is 1 and has constraint value > threshold at the cell, assign 1, indicating developed.

        * Write currentDev to a raster file with header, which is the current development layer.
* Mask out the no value data in the **constraint/suitability layer**.
* Use **MultiCriteriaEval** module's **MaskedWeightedSum** function,

    * Load and read from Attractor table the names of attractor layers
    * Load and read from Attractor layers' weight and normalize
    * Sum all Attractor layers with weights
    * If rval is True, reverse
    * result = constraint layer * summed attractor layer

```python
attractorflag_list = pd.read_csv(table_files['attractors_tbl'])[['layer_name','reverse_polarity_flag']].values.tolist()
mask_layer = np.loadtxt(raster_files['zone_id_ras'], skiprows=6)
for i in range(num_attractors):
    attractor_path = os.path.join(swap_path, attractorflag_list[i][0])
    rev_attractor_flag = attractorflag_list[i][1]
    attractor_layer = np.loadtxt(attractor_path, skiprows=6)
    if rev_attractor_flag == 0:
        standarised_attractor_layer = rt.Standardise(attractor_layer, mask_layer, nodatavalue)
    elif rev_attractor_flag == 1:
        standarised_attractor_layer = rt.RevPolarityStandardise(attractor_layer, mask_layer, nodatavalue)
    attractor_output_path = os.path.join(output_path, 'std_' + attractorflag_list[i][0])
    with open(attractor_output_path, 'w') as f:
        f.write(''.join(lines[:6]))
        np.savetxt(f, standarised_attractor_layer, fmt='%1.3f')
```

```python
rt.RasteriseAreaThresholds(swap_path, header_values, lines[:6], raster_files['constraint_ras'], 
                               raster_files['current_dev_ras'], table_files['constraints_tbl'], num_constraints, parameters['coverage_threshold'])
```
